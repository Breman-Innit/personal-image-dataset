## 2. Linear Regression

Linear regression predicts a continuous target from one or more features by fitting a line or hyperplane. For one feature:

$$\hat{y} = b_0 + b_1x$$

The model learns coefficients by minimizing the sum of squared errors:

$$\mathrm{SSE} = \sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$

Training fits the coefficients using training data. Predictions apply the learned equation to new features. Evaluation should use unseen test data; common metrics are MAE, MSE, RMSE, and $R^2$.

```python
# Training, prediction, and evaluation
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

X = np.array([[1], [2], [3], [4], [5]], dtype=float)
y = np.array([2, 4, 5, 8, 10], dtype=float)
model = LinearRegression().fit(X, y)
predictions = model.predict(X)

rmse = np.sqrt(mean_squared_error(y, predictions))
print("coefficient:", model.coef_[0])
print("intercept:", model.intercept_)
print("prediction for x=6:", model.predict([[6]])[0])
print("MAE:", mean_absolute_error(y, predictions))
print("RMSE:", rmse)
print("R^2:", r2_score(y, predictions))
```

### Linear regression summary
Regression estimates numeric values. Always separate training and evaluation data when measuring generalization, and inspect residuals for systematic errors.

## 3. Confusion Matrix

For binary classification, compare predicted labels with actual labels:

| Actual / Predicted | Positive | Negative |
|---|---:|---:|
| Positive | TP | FN |
| Negative | FP | TN |

- **TP**: correctly predicted positive; **TN**: correctly predicted negative.
- **FP**: negative predicted as positive; **FN**: positive predicted as negative.
- Accuracy = $(TP + TN)/(TP + TN + FP + FN)$
- Precision = $TP/(TP + FP)$
- Recall = $TP/(TP + FN)$
- F1-score = $2 \cdot \mathrm{precision} \cdot \mathrm{recall}/(\mathrm{precision} + \mathrm{recall})$

Example: with TP=8, TN=7, FP=2, FN=3, accuracy is 15/20 = 0.75, precision is 8/10 = 0.80, recall is 8/11 ≈ 0.73, and F1 ≈ 0.76.

```python
# Compute classification metrics for the example
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, precision_score, recall_score

actual = [1] * 11 + [0] * 9
predicted = [1] * 8 + [0] * 3 + [1] * 2 + [0] * 7
print(confusion_matrix(actual, predicted))
print("accuracy:", accuracy_score(actual, predicted))
print("precision:", precision_score(actual, predicted))
print("recall:", recall_score(actual, predicted))
print("F1:", f1_score(actual, predicted))
```

### Confusion matrix summary
Accuracy can hide poor performance on imbalanced data. Precision matters when false positives are costly; recall matters when false negatives are costly; F1 balances precision and recall.

## 4. Object Detection

Object detection finds object instances and classifies them in an image. Each detection commonly includes a class label, confidence score, and bounding box. A box can be represented as `(x_min, y_min, x_max, y_max)` or as `(x_center, y_center, width, height)`.

**IoU (Intersection over Union)** measures box overlap:

$$\mathrm{IoU} = \frac{\mathrm{area\ of\ intersection}}{\mathrm{area\ of\ union}}$$

Ultralytics YOLO is a practical family of real-time vision models for detection and related tasks. A typical workflow loads a model, runs inference, and inspects or saves the results.
